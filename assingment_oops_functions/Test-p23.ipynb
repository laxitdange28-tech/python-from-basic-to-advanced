## Q1 — `if / elif / else` &nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;[20 marks]

Given a student's `marks`, print their grade:

| Marks | Grade |
|-------|-------|
| ≥ 90  | A     |
| 75–89 | B     |
| 60–74 | C     |
| 40–59 | D     |
| < 40  | F     |

Test with `marks = 95`, `marks = 72`, `marks = 35`.  
**Expected output:**
```
Grade: A
Grade: C
Grade: F
```

````python
# Q1 — Write your code here
marks = int(input("Enter a marks : "))

if marks >= 90:
    print("a")
elif marks >=75:
    print("B")
elif marks >=60:
    print("C")
elif marks >= 40:
    print("D")
else:
    print("Fail")    

````

## Q2 — `for` loop &nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;[20 marks]

Loop through the list below and print only the **even numbers**.

```python
numbers = [3, 8, 15, 22, 7, 40, 13, 6, 9, 18]
```

**Expected output:**
```
8
22
40
```

## Q3 — `while` loop                                                                    [20 marks]

Start with `a = 0` and `b = 1`.

**Expected output:**

0
1
1
2
3

8


````python
# Q3 — Write your code here
i = 1
a = 0
b = 1
while i<=8:

    c = a +b
    print(a,end=" ")
    a =b
    b=c
    i +=1
````

## Q5 — Combined: Function + for + if/else &nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;[20 marks]

Write a function `check_pass(scores)` that loops through a list of marks and prints each score as `Pass` (≥ 50) or `Fail`.

Call with:
```python
scores = [72, 45, 88, 30, 55, 49, 91]
```
**Expected output:**
```
72 → Pass
45 → Fail
88 → Pass
30 → Fail
55 → Pass
49 → Fail
91 → Pass
```

````python
# Q5 — Write your code here
scores = [72, 45, 88, 30, 55, 49, 91]

def check_pass(s):
    for i in s:
        if i >= 50:
            print(f"{i} -> pass")
        else:
            print(f"{i} ->Fail")    

check_pass(scores)
````
