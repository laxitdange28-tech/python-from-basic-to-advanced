# Q1. Reverse Each Word in a String

## Problem Statement
Write a Python program to take a sentence from the user and reverse each word individually.

## Sample Input
```python
Enter sentence: hello world
```

## Sample Output
```python
olleh dlrow
```


````python
# Write your code here

# s = input("Enter a string :- ")
s = "Hello world"
words = s.split()

for i in words:
    print(i[::-1],end=" ")

````

# Q2. Find Frequency of Characters

## Problem Statement
Write a Python program to count the frequency of each character in a string.

## Sample Input
```python
Enter string: apple
```

## Sample Output
```python
a : 1
p : 2
l : 1

```

# Q3. Remove Duplicate Elements from List

## Problem Statement
Write a Python program to remove duplicate elements from a list.

## Sample Input
```python
Enter list: [1,2,2,3,4,4,5]
```

## Sample Output
```python
[1, 2, 3, 4, 5]
```


````python
# Write your code here
l = [1,2,2,3,4,4,5]
new_list = []

for i in l:
    if i not in new_list:
        new_list.append(i)

print(new_list)
````

# Q4. Find Second Largest Number in List

## Problem Statement
Write a Python program to find the second largest element in a list.

## Sample Input
```python
Enter list: [10, 20, 5, 40, 30]
```

## Sample Output
```python
30
```


````python
# Write your code here

l = [10,20,5,40,30]

max_element = float('-inf')

for i in l:
    if i > max_element:
        max_element = i

print(max_element)        
````

# Q5. Merge Two Dictionaries

## Problem Statement
Write a Python program to merge two dictionaries into one.

## Sample Input
```python
dict1 = {'a':1, 'b':2}
dict2 = {'c':3, 'd':4}
```

## Sample Output
```python
{'a': 1, 'b': 2, 'c': 3, 'd': 4}
```


````python
# Write your code here

dict1 = {'a':1,'b':2}
dict2 ={'c':3 , 'd':4}

for i in dict2:
    if i not in dict1:
        dict1[i] = dict2[i]
        
print(dict1)         
````

# Q9. Create Dictionary from Two Lists

## Problem Statement
Write a Python program to create a dictionary using two lists.

## Sample Input
```python
keys = ['name', 'age']
values = ['Shiva', 22]
```

## Sample Output
```python
{'name': 'Shiva', 'age': 22}
```


````python
# Write your code here
keys = ['name','age']
values = ['shiva',22]
d = {}

for i in range(len(keys)):
    d[keys[i]] = values[i]

print(d)    
````

# Q10. Find Longest Word in String

## Problem Statement
Write a Python program to find the longest word in a sentence.

## Sample Input
```python
Enter sentence: Python programming is amazing
```

## Sample Output
```python
programming
```


````python
# Write your code here

sentence = input("Enter sentence: ")


words = sentence.split()

longest = words[0]

for word in words:

    if len(word) > len(longest):
        longest = word

print(longest)
````
