# Q1. Create a Player Class

## Problem Statement
Create a class named `Player` using constructor.

Store:
- Player Name
- Jersey Number

Create a method to display player details.

Create two objects.

## Sample Output
```python
Player Name: Virat
Jersey Number: 18

Player Name: Rohit
Jersey Number: 45
```


````python
# Write your code her
class Player:
    def __init__(self,player,jersey):
        self.player = player
        self.jersey = jersey

    def show(self):
        print(f"{self.player}  {self.jersey}")

obj1 = Player("virat", 19)
obj1.show()
obj2 = Player("Rohit" , 45)
obj2.show()



    
````

# Q3. Employee Bonus Calculator

## Problem Statement
Create a class named `Employee`.

Store:
- Employee Name
- Salary

Create method to calculate bonus:
- Bonus = 10% of salary

## Sample Output
```python
Employee Name: Aman
Salary: 50000
Bonus: 5000
```


````python
# Write your code here
class Employee:
    def __init__(self,name,salary):
        self.name = name
        self.salary = salary

    def bonus(self):
        Bonus = self.salary*10/100
        print(f"Bonus : {Bonus}")

    def show(self):
        print(f"Employee Name : {self.name} \n Salary : {self.salary}")   

obj = Employee("laxit",50000)             
obj.show()
obj.bonus()
````

# Q4. Circle Class

## Problem Statement
Create a class named `Circle`.

Use constructor to take radius.

Create methods to calculate:
- Area
- Circumference

Use:
```python
3.14 * r * r
2 * 3.14 * r
```

## Sample Output
```python
Area: 78.5
Circumference: 31.400000000000002
```


````python
# Write your code here
class Circle:
    def __init__(self,radius):
        self.radius = radius

    def area(self):
        Area = 3.14 * self.radius *self.radius
        print(f"Area : {Area}")

    def  circumference(self):
        print(f"Circumference : {2*3.14*self.radius}")    

obj = Circle(5)
obj.area()       
obj.circumference()
````

# Q10. Food Item Class

## Problem Statement
Create a class named `FoodItem`.

Store:
- Food Name
- Price
- Quantity

Create method to calculate total bill.

Formula:
```python
Total = Price * Quantity
```

## Sample Output
```python
Food Name: Pizza
Total Bill: 800
```


````python
# Write your code here
class FoodItem():
    def __init__(self,foodName,price,quantity):
        self.foodName = foodName
        self.price = price
        self.quantity = quantity

    def tatalbill(self):
        print(f"Food Name: {self.foodName} \n Total Bill : {self.price*self.quantity}")

foods = FoodItem("Pizza",800,1)
foods.tatalbill()            
    
````
