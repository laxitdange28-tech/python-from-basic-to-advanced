## 1. Print All Values using *args

### Problem Statement
Write a function using *args to print all values passed to the function.

### Example Input
```python
10, 20, 30
```

### Example Output
```python
10 20 30
```


````python
"""Method - 1"""
# def show(*args):
#     return args

# print(show(10,20,30)) 

"""Method - 2"""

def print_value(*args):

    for value in args:
        print(value,end=" ")

print_value(10,20,30)        
````


## 2. Sum of Numbers using *args

### Problem Statement
Write a function using *args to find the sum of numbers.

### Example Input
```python
1, 2, 3
```

### Example Output
```python
6
```


````python
def sum_of_value(*args):
    sum = 0

    for value in args:
        sum += value

    print(sum)    

sum_of_value(1,2,3)    
````


## 3. Count Arguments using *args



## 4. Print Student Details using **kwargs

### Problem Statement
Write a function using **kwargs to print student details.

### Example Input
```python
name='Parth', age=22
```

### Example Output
```python
name : Parht
age : 22
```


````python
def student_details(**kwargs):
    
    for key,value in kwargs.items():
        print(f"{key} : {value}")

student_details(name = "parth",age = 22)
````
