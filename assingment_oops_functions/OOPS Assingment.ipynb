Q1. Student Information System
      
    Problem Statement

    Create a class named Student.

    Store:

      Student Name
      Roll Number
      Course

Create a method display_details() to display all student information.

       Output:
       Student Name : Rahul
       Roll Number  : 101
       Course       : Python

```python
#write your code 
class Student:
    def __init__(self,name,rollNumber,course):
        self.name = name
        self.rollNumber = rollNumber
        self.course = course

    def show(self):
        print(f"Student Name : {self.name}\nRoll Number: {self.rollNumber}\nCourse : {self.course}")    

student1 = Student("Rahul",101,"Python")
student1.show()        
```

Q3. Bank Account System

    Problem Statement

    Create a class named BankAccount.
 
    Store:

    Account Holder Name
    Balance

    Create methods:

     deposit()
     withdraw()
    check_balance()
    
    Output: Current Balance = 11000

```python
#write your code 
class BankAccount:
    def __init__(self,account_holder_name,balance):
        self.account_holder_name = account_holder_name
        self.balance = balance

    def deposit(self,balance):  
        self.balance += balance
        print(f"Deposit Blance : {self.balance}")

    def withdraw(self,balance):
        if balance <= self.balance:
            self.balance -= balance
            print(f"Withdraw Blance : {self.balance}")

    def balance_check(self):
        print(f"Total Balance : {self.balance} ")  


obj = BankAccount("Laxit",1000)
obj.deposit(500)
```
