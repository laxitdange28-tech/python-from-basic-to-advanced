# Q1. Find Minimum Number using *args

## Problem Statement
Write a Python function using *args to find the minimum number from given arguments.

## Sample Function Call
```python
minimum(10, 5, 30, 2, 8)
```

## Sample Output
```python
2
```


````python
# Write your code here
def minimum(*args):
    min_value = float('inf')

    for i in args:
        if i < min_value:
            min_value = i
    return min_value

print(minimum(10,5,30,2,8))        
````

# Q2. Sum of Even Numbers using *args

## Problem Statement
Write a Python function using *args to calculate the sum of all even numbers.

## Sample Function Call
```python
even_sum(1,2,3,4,5,6)
```

## Sample Output
```python

```

# Q3. User Profile Generator using **kwargs

## Problem Statement
Write a Python function using **kwargs to print a user profile dynamically.

## Sample Function Call
```python
profile(name='Aman', city='Delhi', skill='Python')
```

## Sample Output
```python
name : Aman
city : Delhi
skill : Python
```


````python
# Write your code here
def profile(**kwargs):

    for key , value in kwargs.items():
        print(f"{key} : {value}")

profile(name="Aman",city="Dehli",skil="Python")       
````

# Q4. Count Keyword Arguments using **kwargs

## Problem Statement
Write a Python function using **kwargs to count total keyword arguments passed.

## Sample Function Call
```python
details(name='Arjun', age=25, course='DS')
```

## Sample Output
```python
3
```


````python
# Write your code here
def count_keyword(**kwargs):
    count = 0
    for key,value in kwargs.items():
        count += 1
    return count
print(count_keyword(name="Arjun",age="25",course="DS"))
````

# Q6. Sort Dictionary by Value using Lambda

## Problem Statement
Write a Python program to sort a dictionary based on values using lambda.

## Sample Input
```python
{'a':3, 'b':1, 'c':2}
```

## Sample Output
```python
[('b',1), ('c',2), ('a',3)]
```


````python
# Write your code here
d = {'a':3,'b' : 1 ,'c' : 2}
d_sort = sorted(d.items(),key=lambda x:x[1])

print(d_sort)
````

# Q7. Filter Odd Numbers using Lambda

## Problem Statement
Write a Python program using lambda and filter() to extract odd numbers from a list.

## Sample Input
```python
[1,2,3,4,5,6,7]
```

## Sample Output
```python
[1,3,5,7]
```


````python
# Write your code here
l = [1,2,3,4,5,6,7]
odd_number = list(filter(lambda x: x%2 != 0 , l))
print(odd_number) 
````

# Q9. Student Marks Analyzer using *args

## Problem Statement
Write a Python function using *args to calculate:
- Total Marks
- Average Marks
- Highest Marks

## Sample Function Call
```python
marks(70,80,90,60)
```

## Sample Output
```python
Total: 300
Average: 75.0
Highest: 90
```


````python
# Write your code here
def mark_Analyzer(*args):
    print(f"Total",sum(args))
    print(f"Average:{sum(args)/len(args)}")
    print(f"Highest : {max(args)}")

mark_Analyzer(70,80,90,60)    
````

# Q10. Dynamic Bill Generator using **kwargs

```python
bill(Shirt=1200, Shoes=2500, Watch=3000)
```

## Sample Output
```python
Shirt : 1200
Shoes : 2500
Watch : 3000

```


````python
# Write your code here
def bill(**kwargs):

    for key , value in kwargs.items():
        print(f"{key} : {value}")
    
    
    print(f"Total Bill : {sum(kwargs.values())}")    

bill(Shirt=1200,shoes=2500,watch=3000)
    
````
