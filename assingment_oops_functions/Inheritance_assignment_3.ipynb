# Q1. Electronics and Laptop Class

## Problem Statement
Create a parent class `Electronics` and child class `Laptop`.

Store:
- brand_name
- processor

Display all details using method.

## Sample Input
```python
Brand: Dell
Processor: i7
```

## Sample Output
```python
Brand: Dell
Processor: i7
```


````python
# Write your code here
class Electronics:
    def __init__(self,brand_name):
        self.brand_name = brand_name
        
class Laptop(Electronics):
    def __init__(self, brand_name,processor):
        super().__init__(brand_name)
        self.processor = processor

    def show(self):
        print(f"Brand: {self.brand_name}\nProcessor: {self.processor}")
        
obj = Laptop("Dell","i7")
obj.show()        
````

```python
Name: Arjun

Specialization: Cardiologist
```

## Sample Output

````python
# Write your code 

class Human:
    def __init__(self,name,age):
        self.name = name
        self.age = age


class Doctor(Human):
    def __init__(self, name, age,specialization):
        super().__init__(name, age) 
        self.specialization = specialization

    def show(self):
        print(f"Name: {self.name}\nAge: {self.age}\nSpecialization: {self.specialization}")

d = Doctor("Arjun",35,"Cardiologist")
d.show()        
````


# Q3. Country, State and City Class

## Problem Statement
Create multilevel inheritance using:
- Country
- State
- City

Store:
- country_name
- state_name
- city_name

Display all details.

## Sample Input
```python
Country: India
State: Madhya Pradesh
City: Bhopal
```

## Sample Output
```python
Country: India
State: Madhya Pradesh
City: Bhopal
```


````python
# Write your code here

class Country:
    def __init__(self,country_name):
        self.country_name = country_name

class State(Country):
    def __init__(self, country_name,state_name):
        super().__init__(country_name)
        self.state_name = state_name

class City(State):
    def __init__(self, country_name, state_name,city_name):
        super().__init__(country_name, state_name)
        self.city_name = city_name

    def show(self):
        print(f"Country: {self.country_name}\nState: {self.state_name}\nCity: {self.city_name}")

c1 = City("India","Madhya Pradesh","Bhopal")
c1.show()

            
````


# Q5. Father and Mother Multiple Inheritance

## Problem Statement
Create:
- Father class
- Mother class
- Son class

Store:
- father_profession
- mother_profession
- son_name

Display all details.

## Sample Input
```python
Father Profession: Engineer
Mother Profession: Teacher
Son Name: Rahul
```

## Sample Output
```python
Father Profession: Engineer
Mother Profession: Teacher
Son Name: Rahul
```


````python
# Write your code here
class Father:
    def __init__(self,father_profession):
        self.father_profession = father_profession

class Mother:
    def __init__(self,mother_profession):
        self.mother_profession = mother_profession

class Son(Father,Mother):
    def __init__(self, father_profession,mother_profession,son_name):
        Father.__init__(self,father_profession)  
        Mother.__init__(self,mother_profession)      
        self.son_name = son_name

    def show(self):
        print(f"Father Profession : {self.father_profession}\nMoother Proofession : {self.mother_profession}\nSon Name : {self.son_name}")

s = Son("Engineer","Teacher","Rahul")
s.show()        

````


# Q10. Teacher Hierarchical Inheritance

## Problem Statement
Create hierarchical inheritance using:
- Teacher
- MathsTeacher
- ScienceTeacher

Store:
- teacher_name
- subject

Display details using both child classes.

## Sample Input
```python
Teacher: Neha
Subject: Maths

Teacher: Rahul
Subject: Science
```

## Sample Output
```python
Teacher: Neha
Subject: Maths

Teacher: Rahul
Subject: Science
```


````python
# Write your code here
class Teacher:
    def __init__(self,teacher_name,subject):
        self.teacher_name = teacher_name
        self.subject = subject

class MathTeacher(Teacher):
    def show(self):
        print(f"Teacher: {self.teacher_name}\nSubject: {self.subject}\n")
class ScinceTeacher(Teacher):
   
    def show(self):
        print(f"Teacher: {self.teacher_name}\nSubject: {self.subject}")    

m = MathTeacher("Neha","Maths")
m.show()

s = ScinceTeacher("Rahul","Science")
s.show()
    
````
