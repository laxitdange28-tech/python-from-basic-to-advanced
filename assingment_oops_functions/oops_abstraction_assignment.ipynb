# Q1. Animal Sound Program

## Problem Statement
Create an abstract class `Animal` with abstract method `sound()`.

Create child class `Dog` and print dog sound.

## Sample Input
```python
Dog
```

## Sample Output
```python
Dog barks
```


````python
from abc import ABC , abstractmethod
class Animal(ABC):
    @abstractmethod
    def sound():
        pass

class Dog(Animal):
    def __init__(self,name):
        self.name = name
        
    def sound(self):
        print(f"{self.name} is barke ")

d = Dog("Dog")
d.sound()
````

## Sample Input
```python
BMW
```

## Sample Output


# Q3. Shape Area Program

## Problem Statement
Create abstract class `Shape` with abstract method `area()`.

Create child class `Square`.

Calculate square area.

## Formula
```python
side * side
```

## Sample Input
```python
Side: 5
```

## Sample Output
```python
Area: 25
```


````python
# Write your code here
from abc import ABC , abstractmethod

class Shape(ABC):
    @abstractmethod
    def area():
        pass

class Square(Shape):
    def __init__(self,side):
        self.side = side

    def area(self):
        print(f"Area : {self.side * self.side}")

obj =  Square(5)   
obj.area()       
        
````

Create abstract class `Employee` with abstract method `work()`.

Create child class `Developer`.

Print work message.

## Sample Input
```python
Python Developer
```

## Sample Output

````python
# Write your code here
from abc import ABC,abstractmethod

class Employee(ABC):
    @abstractmethod
    def work():
        pass

class Developer(Employee):
    def __init__(self,name):
        self.name = name

    def work(self):
        print(f"{self.name} Developer")


obj = Developer("Python")
obj.work()
````
