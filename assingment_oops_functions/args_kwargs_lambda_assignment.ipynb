# Q1. Sum of Multiple Numbers using *args

## Problem Statement
Write a Python function using *args to calculate the sum of all numbers passed as arguments.

## Sample Function Call
```python
addition(10, 20, 30)
```

## Sample Output
```python
60
```


````python
# Write your code here
def addition(*args):
    return sum(args)

print(addition(10,20,30))
````

# Q2. Find Maximum Number using *args

## Problem Statement
Write a Python function using *args to find the largest number.

## Sample Function Call
```python
maximum(5, 20, 15, 50, 10)
```

## Sample Output
```python
50
```


````python
# Write your code here
def maximum(*args):
    max_number = float('-inf')
    for i in args:
        if i > max_number:
            max_number = i

    return max_number

print(maximum(5,20,15,50,10))        
````

# Q3. Print Student Details using **kwargs

## Problem Statement
Write a Python function using **kwargs to print student information.

## Sample Function Call
```python
student(name='Omprakash', age=22, course='Python')
```

## Sample Output
```python
name : Omprakash
age : 22
course : Python
```


````python
# Write your code here
def student_info(**kwargs):
    for key , value in kwargs.items():
        print(f"{key} : {value}")

student_info(name="omprakash",age=22,course="Python")
````

# Q4. Count Total Arguments using *args

## Problem Statement
Write a Python function using *args to count how many arguments are passed.

## Sample Function Call
```python
count_args(1, 2, 3, 4, 5)
```

## Sample Output
```python

```


````python
# Write your code here
def counts(*args):
    count = 0
    for i in args:
        count +=1
    return count

print(counts(1,2,3,4,5))    
````

```python

```
