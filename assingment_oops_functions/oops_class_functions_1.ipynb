# Q1. Create a Student Class

## Problem Statement
Create a class named `Student` and create an object of that class.

Print:
- Student Name
- Student Age

## Sample Output
```python
Name: Raj
Age: 17
```


````python
# Write your code here
class Student:
    def __init__(self,name,age):
        self.name = name
        self.age = age
    def show(self):
        print(f" Name: {self.name} \n Age: {self.age}")

obj = Student("Laxit",19)
obj.show()


````
